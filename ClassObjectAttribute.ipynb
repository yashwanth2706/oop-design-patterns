```python

class Fruit():
    
    # Class Object Attribute
    # Same for any instance of the class
    
    Organic = True # Not connected to any instance
    # Dry is always T
    
    def __init__(self, name):
        self.name = name
        
    # Operations/ Actions -> Methods
    def identify(self):
        print(f"I'm {self.name}")
        
    def is_organic(self):
        return self.Organic
```

```python
my_fruit = Fruit('Apple')
```

```python
my_fruit.Organic
```

```python
my_fruit.identify
```

```python
my_fruit.identify()
```

```python
my_fruit.is_organic()
```

```python
class Circle():
    # Class Object Attribute
    pi = 3.14 # Is always True for this Class
    
    def __init__(self, radius=1):
        
        self.radius = radius
        self.area = radius*radius*self.pi
    
    # Method
    def get_circumference(self):
        return self.radius * self.pi * 2
```

```python
wheel = Circle(30)
```

```python
wheel.pi
```

```python
wheel.radius
```

```python
wheel.get_circumference()
```

```python
wheel.area
```
