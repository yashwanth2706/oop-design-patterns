```python
class Fruit():
    def __init__(self, name):
        # Attributes
        # We take in the argument
        # Assign it using self.attribute_name
        self.name = name
```

```python
new_fruit = Fruit()
```

```python
new_fruit = Fruit("Orange")
```

```python
new_fruit.name
```

```python
class Fruit():
    def __init__(self, my_name):
        # Attributes
        # We take in the argument
        # Assign it using self.attribute_name
        self.my_atrribute = my_name
```

```python
class Car():
    def __init__(self, name, color, nitro):
        self.name = name
        self.color = color
        # Expect boolean True/False
        self.nitro = nitro
```

```python
new_car = Car(name='F1', color='red', nitro=False)
```

```python
type(new_car)
```

```python
new_car.name
```

```python
new_car.color
```

```python
new_car.nitro
```
