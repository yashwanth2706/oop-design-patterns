```python
class Animal():
    
    def __init__(self):
        print("Animal Created")
        
    def who_am_i(self):
        print("I am an Animal")
        
    def eat(self):
        print("I'm Eating")
```

```python
class Dog(Animal):
    
    def __init__(self):
        Animal.__init__(self)
        print("Dog Created")
        
    def eat(self):
        print("I'm a dog and I'm eating")
        
    def bark(self):
        print("WOOF!")
```

```python
mydog = Dog()
```

```python
mydog.eat()
```

```python
mydog.who_am_i()
```

```python
mydog.bark()
```

```python
class Dog():
    
    def __init__(self, name):
        self.name = name
    
    def speak(self):
        return self.name + "Says Woof!"
```

```python
class Cat():
    def __init__(self, name):
        self.name = name
        
    def speak(self):
        return self.name + "Says Meow"
```

```python
niko = Dog("niko")
felix = Cat("felix")
```

```python
niko.speak()
```

```python
felix.speak()
```

```python
for pet in [niko, felix]:
    print(type(pet))
    print(type(pet.speak()))
```

```python
def pet_speak(pet):
    print(pet.speak())
```

```python
pet_speak(niko)
```

```python
pet_speak(felix)
```

```python
# Abstract Class only used as a Based Class
class Animal():
    def __init__(self, name):
        self.name = name
        
    def speak(self):
        raise NotImplementedError("Subclass must implement this abstract method")
```

```python
myanimal = Animal('fred')
```

```python
myanimal.speak()
```

```python
class Dog(Animal):
    def speak(self):
        return self.name + "Says Meow!"
```

```python
fido = Dog("Fido")
```

```python
isis = Cat("Isis")
```

```python
fido.speak()
```

```python
isis.speak()
```
